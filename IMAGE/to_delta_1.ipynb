```python
import numpy as np
import pandas as pd
```

```python
file1 = "SUBGRID_DATA_IMAGEBASED_DELTA_100.txt"
file2 = "SUBGRID_DATA_IMAGEBASED_100.txt"
```

```python
df1 = pd.read_csv(file1, sep="\t")
df2 = pd.read_csv(file2, sep="\t")
```

```python
df1.head()
```

```python
df2.head()
```

```python
df = pd.merge(df1, df2, on=['IMAGE_NAME', 'PT_LAT', 'PT_LON', 'BUILD_PROB_OLD'], 
              how='outer', indicator=True)
df.head()
```

```python
headers_delta = list(df.columns.values)
print(headers_delta)
```

```python
df_delta = df[['IMAGE_NAME', 
                'PT_LAT', 
                'PT_LON', 
                'BUILD_PROB_OLD', 
                'BUILD_PROB_NEW_x', 
                'BUILD_PROB_NEW_y', 
                'BUILD_PROB_delta_x', 
                'BUILD_PROB_delta_y']]
#df_delta.head()
```

```python
df_delta = df_delta.rename(columns={'BUILD_PROB_NEW_x': 'BUILD_PROB_NEW', 
                                    'BUILD_PROB_NEW_y': 'BUILD_PROB_NEW_100', 
                                    'BUILD_PROB_delta_x': 'BUILD_PROB_delta', 
                                    'BUILD_PROB_delta_y': 'BUILD_PROB_delta_100'})
df_delta.head()
```

```python
df_delta.to_csv('SUBGRID_DATA_IMAGEBASED_DELTA_100_EXT', sep='\t', index=False)
```
