```python
import numpy as np
import pandas as pd
```

```python
file1 = "SUBGRID_DATA_IMAGEBASED_FOLIAGE_NEW.txt"
file2 = "SUBGRID_DATA_IMAGEBASED_FOLIAGE_103.txt"
```

```python
df1 = pd.read_csv(file1, sep="\t")
df2 = pd.read_csv(file2, sep="\t")
```

```python
df1.head()
```

```python
df2.head()
```

```python
headers_delta = list(df2.columns.values)
headers_delta = headers_delta[1:]
print(headers_delta)
```

```python
headers_delta[-1] = 'FOLIAGE_PROB_103'
headers_delta.append('FOLIAGE_PROB_NEW')
headers_delta.append('FOLIAGE_PROB_delta_103')
print(headers_delta)
```

```python
df = pd.merge(df1, df2, on=['Unnamed: 0', 'IMAGE_NAME', 'PT_LAT', 'PT_LON'], how='outer', indicator=True)
df.head()
```

```python
#df.set_index(['PT_LAT', 'PT_LON'])
```

```python
list_delta = []
for i in df['Unnamed: 0'].values:
    FOLIAGE_PROB_delta = df['FOLIAGE_PROB_y'][i] - df['FOLIAGE_PROB_x'][i]
    list_delta.append([df['IMAGE_NAME'][i], df['PT_LAT'][i], df['PT_LON'][i], df['FOLIAGE_PROB_y'][i], df['FOLIAGE_PROB_x'][i], FOLIAGE_PROB_delta])
```

```python
df_delta = pd.DataFrame(list_delta, columns=headers_delta)
```

```python
df_delta.head()
```

```python
df_delta.to_csv('SUBGRID_DATA_IMAGEBASED_FOLIAGE_DELTA_103.txt', sep='\t', index=False)
```
