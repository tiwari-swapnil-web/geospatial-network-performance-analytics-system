```python
import numpy as np
import pandas as pd
from functools import reduce
```

```python
file1 = "SUBGRID_DATA_IMAGEBASED_FOLIAGE_DELTA_100.txt"
file2 = "SUBGRID_DATA_IMAGEBASED_FOLIAGE_DELTA_101.txt"
file3 = "SUBGRID_DATA_IMAGEBASED_FOLIAGE_DELTA_102.txt"
file4 = "SUBGRID_DATA_IMAGEBASED_FOLIAGE_DELTA_102+OUTDATED.txt"
file5 = "SUBGRID_DATA_IMAGEBASED_FOLIAGE_DELTA_103.txt"
```

```python
df1 = pd.read_csv(file1, sep="\t")
df2 = pd.read_csv(file2, sep="\t")
df3 = pd.read_csv(file3, sep="\t")
df4 = pd.read_csv(file4, sep="\t")
df5 = pd.read_csv(file5, sep="\t")
```

```python
data_frames = [df1,df2,df3,df4,df5]
```

```python
df_merged = reduce(lambda  left,right: pd.merge(left,right,on=['IMAGE_NAME', 'PT_LAT', 'PT_LON', 'FOLIAGE_PROB_NEW'],
                                            how='outer'), data_frames)
df_merged.head()
```

```python
headers_delta = list(df_merged.columns.values)
print(headers_delta)
```

```python
df_delta = df_merged[['IMAGE_NAME', 
                'PT_LAT', 
                'PT_LON', 
                'FOLIAGE_PROB_NEW', 
                'FOLIAGE_PROB_100', 
                'FOLIAGE_PROB_101',
                'FOLIAGE_PROB_102',
                'FOLIAGE_PROB_102+OUTDATED',
                'FOLIAGE_PROB_103',
                'FOLIAGE_PROB_delta_100',
                'FOLIAGE_PROB_delta_101',
                'FOLIAGE_PROB_delta_102',
                'FOLIAGE_PROB_delta_102+OUTDATED',
                'FOLIAGE_PROB_delta_103']]
df_delta.head()
```
