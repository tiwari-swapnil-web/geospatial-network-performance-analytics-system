```python
from tensorflow.keras.models import load_model
model_loaded1 = load_model('building-detector-v11-colored.h5')
```

```python

#2957,2762,2351,2398,2360,2949,2970,3928
#,3054,3081,3354
image_path="E:/IMAGEPRO/geotiff/tiffs/tiffs.49.tif"
#input:image_path
#input : chunk_size=50
#input : img_size = 224

width=500
height=500
chunk_size=50
y_size=chunk_size 
x_size=chunk_size 

img_size = 224
img = cv2.imread(image_path)
imS = cv2.resize(img, (500,500)) 

#####################
imagelist=[]
all_grids=[]
building_grids=[]
other_grids=[]
idx=0
for ys in np.arange(0,height,y_size):
    for xs in np.arange(0,width,x_size):
        if(ys<=height-y_size and xs<=width-x_size):
            #print(ys,xs)
            idx=idx+1
            start_point = (ys,xs)
            end_point = (ys+y_size,xs+x_size)
            
            all_grids.append((xs,ys))
            
            crop_img = imS[ys:ys+y_size, xs:xs+x_size][...,::-1] 
            resized_arr = cv2.resize(crop_img, (img_size, img_size))
            
            imagelist.append(resized_arr)

img_size = 224
imagelist_in = np.array(imagelist) / 255
imagelist_in.reshape(-1, img_size, img_size, 1)

predict_x1 = model_loaded1.predict(imagelist_in)

classes_x=np.argmax(predict_x1,axis=1)


#ARGNAX
for coords,cls,pred in zip(all_grids,classes_x,predict_x1):
    if(cls==0): # and abs(pred[0]-pred[1])>=0.1
        print(pred[0],pred[1])
        xs_=coords[0]
        ys_=coords[1]
        cv2.rectangle(imS, (xs_,ys_), (xs_+x_size,ys_+y_size), (0, 0, 255), 2)
"""
#THESH
for coords,cls,pred in zip(all_grids,classes_x,predict_x1):

    if(cls[0]==0):
        #print(cls)
        xs_=coords[0]
        ys_=coords[1]
        cv2.rectangle(imS, (xs_,ys_), (xs_+x_size,ys_+y_size), (0, 0, 255), 2)        
"""        
cv2.namedWindow("Window")

# Starting The Loop So Image Can Be Shown

wait_time = 1000
while cv2.getWindowProperty('Window', cv2.WND_PROP_VISIBLE) >= 1:
    cv2.imshow("Window",imS)
    keyCode = cv2.waitKey(wait_time)
    if (keyCode & 0xFF) == ord("q"):
        cv2.destroyAllWindows()
        break
```
